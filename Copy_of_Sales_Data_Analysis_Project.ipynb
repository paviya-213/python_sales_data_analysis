```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("All Libraries are ready")
data = {"order_id":list(range(1001,1021)),
        "order_date":["2026-01-05","2026-01-10","2026-01-15","2026-01-20","2026-01-25",
                      "2026-02-02","2026-02-08","2026-02-14","2026-02-18","2026-02-25",
                      "2026-03-03","2026-03-09","2026-03-15","2026-03-19","2026-03-25",
                      "2026-04-04","2026-04-10","2026-04-16","2026-04-22","2026-04-28"
                      ],
        "product":["Laptop","Phone","Mouse","Headphones","Keyboard",
                   "Keyboard","Mouse","Laptop","Phone","Headphones",
                   "Headphones","Laptop","Phone","Mouse","Keyboard",
                   "Mouse","Laptop","Headhones","Keyboard","Phone"],
        "Category":["Electronics","Electronics","Accessories","Accessories",
                    "Accessories","Electronics","Electronics","Accessories",
                    "Accessories","Accessories","Electronics","Electronics",
                    "Accessories","Accessories","Accessories","Electronics",
                    "Electronics","Accessories","Accessories","Accessories"],
        "Quantity":[2,3,5,4,2,2,1,6,3,4,
                    2,4,3,5,2,1,3,4,2,3],
        "Unit_Price":[50000,20000,500,1500,1000,
                      1000,500,50000,20000,1500,
                      1500,50000,20000,500,1000,
                      500,50000,1500,1000,20000],
        "Region":["South","North","East","West","South",
                  "East","North","South","West","North",
                  "South","West","East","North","South",
                  "West","South","East","North","west"]}
df = pd.DataFrame(data)
df.head()
print("Dataset shape:",df.shape)
print("\nColumn names:")
print(df.columns)
print("\nDataset information:")
df.info()
print("\nStatistical Summary:")
display(df.describe(include="all"))
print("Missing Values:")
print(df.isnull().sum())
print("\nDuplicate rows:")
print(df.duplicated().sum())
df.loc[0,"quantity"] = np.nan
df = pd.concat([df,df.iloc[[1]]],ignore_index=True)
print("Missing values after test changes:")
print(df.isnull().sum())
print("Duplicate rows after test changes:")
print(df.duplicated().sum())
df = df.drop_duplicates().copy()
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["order_date"] = pd.to_datetime(df["order_date"])
print("Missing Values after Cleaning:")
print(df.isnull().sum())
print("\nDuplicate rows after cleaning:")
print(df.duplicated().sum())
print("\nCleaned dataset:")
display(df.head())
df["Revenue"] = df["Quantity"]*df["Unit_Price"]
display(df.head(10))
total_orders = df["order_id"].nunique()
total_units = df["Quantity"].sum()
total_Revenue = df["Revenue"].sum()
average_order_value = df.groupby("order_id")["Revenue"].sum().mean()
print("total orders:",total_orders)
print("total units sold:",total_units)
print("total Revenue:₹", round(total_Revenue, 2))
print("average Revenue per order:₹", round(average_order_value, 2))
product_sales = (df.groupby("product").agg(units_sold=("Quantity","sum"), total_Revenue=("Revenue", "sum"))
.sort_values("total_Revenue", ascending=False)
)
display(product_sales)
top_product = product_sales["total_Revenue"].idxmax()
top_revenue = product_sales["total_Revenue"].max()
print("top product by revenue:",top_product)
print("revenue generated:₹",top_revenue)
df["Month"] = df["order_date"].dt.to_period("M").astype(str)
monthly_sales = (df.groupby("Month")["Revenue"].sum().reset_index())
display(monthly_sales)
best_month = monthly_sales.loc[monthly_sales["Revenue"].idxmax()]
print("best sales month:", best_month["Month"])
print("Revenue:₹", best_month["Revenue"])
Category_sales = (df.groupby("Category")["Revenue"].sum().sort_values(ascending=False))
display(Category_sales)
region_sales = (df.groupby("Region")["Revenue"].sum().sort_values(ascending=False))
display(region_sales)
product_sales["total_Revenue"].sort_values().plot(kind="bar",figsize=(8,5),title="Revenue by product")
plt.xlabel("Product")
plt.ylabel("Revenue(₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
plt.figure(figsize=(9,5))
plt.plot(monthly_sales["Month"],monthly_sales["Revenue"],marker="o")
plt.title("monthly sales Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue(₹)")
plt.grid(True)
plt.tight_layout()
plt.show()
plt.figure(figsize=(7,5))
sns.barplot(x=region_sales.index,y=region_sales.values)
plt.title("Revenue by Redion")
plt.xlabel("Region")
plt.ylabel("Revenue(₹)")
plt.tight_layout()
plt.show()
print("SALES ANALYSIS REPORT")
print("-"*30)
print(f"Total orders:{total_orders}")
print(f"Total units sold:{total_units}")
print(f"Total Revenue:₹{total_Revenue:,.2f}")
print(f"Average Revenue per order:₹{average_order_value:,.2f}")
print(f"Top product bu Revenue:{top_product}"
f"(₹{top_revenue:,.2f}")
print(f"Best_month:{best_month["Month"]}"
f"(₹{best_month['Revenue']:,.2f}")
print(f"Top region by Revenue:{region_sales.idxmax()}"
f"(₹{region_sales.max():,.2f}")
df.to_csv("cleaned_sales_data.csv",index=False)
product_sales.to_csv("product_sales_summary.csv")
monthly_sales.to_csv("monthly_sales_summary.csv")
region_sales.to_csv("region_sales_summary.csv")
print("All CSV files have been created!")
from google.colab import files
files.download("cleaned_sales_data.csv")

```
